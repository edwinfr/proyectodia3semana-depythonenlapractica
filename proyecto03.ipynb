```python
import yfinance as yf

tikers = ["AAPL", "NVDA", "MSFT"]
datos = yf.download(tikers, period="1y")

precios = datos["Close"]
precios.head()
```

```python
precios.tail()
```

## Estadisticas simples

```python
precios.describe()
```

## Graficos Interactivos

```python
import plotly.express as px
```

```python
px.line(precios,
        title="Cotización Historica:Apple, Microsoft y Nvidia (ultimos 12 meses)",
        labels={"value": "Precio (USD)", "Date": "Fecha"},
        template="plotly_dark")
```

```python
px.bar(precios.mean(),
title="Precio promedio: Apple, Nvidia y Microsoft(ultimos 12 meses)",
labels={"value": "Precio promedio (USD)", "index":
"Empresa"},

color=precios.mean().index,
template="plotly_dark")
```

```python
fig.write_html("Inversion de 1000 USD.html")
```

## Informe con IA

```python
import os
from getpass import getpass
from google import genai as gemini

api_key = os.getenv("GEMINI_API_KEY") or getpass(
    "Escribe tu clave de Gemini (no se mostrará): "
)
cliente = gemini.Client(api_key=api_key)
prompt = f"""
Haz un análisis del último año de las acciones de las
empresas Apple, Microsoft y Nvidia considerando eventos y
noticias que justifiquen las alzas y bajas relevantes de
estas empresas.
{precios}
"""
respuesta = cliente.models.generate_content(
    model="gemini-3.5-flash",
    contents=prompt)
```
